```python
# binary search
a = list(map(int,input().split(" ")))
target = int(input())
n=len(a)
start = 0
end =n-1
idx=-1
while start<=end:
  mid =(start+end)//2
  if a[mid] == target:
    idx=mid
    break
  elif target<a[mid]:
    end =mid-1
  elif target>a[mid]:
    start=mid+1

if idx == -1 :
  print("target not found")
else :
  print("target found at index", idx)

```

```python
# Selection sort
l = list(map(int,input().split(" ")))
n=len(l)
for j in range(0,n-1):
    minn=l[j]
    mIdx=j
    for i in range(j+1,n):
        if l[i]<minn:
            minn=l[i]
            mIdx=i
    temp=l[j]
    l[j]=l[mIdx]
    l[mIdx]=temp
    # print(l)
print(l)
```

```python
# finding max using recursion

def findMax(l,idx):
    if idx==len(l)-1:
        return l[idx]
    maxx=l[idx]
    res=findMax(l,idx+1)
    if res>maxx:
        return res
    else:
        return maxx
l=[2,1,5,8,10]
a=findMax(l,0)

print(a)
```

```python
# finding min using recursion

def findMin(l,idx):
    if idx==len(l)-1:
        return l[idx]
    minn=l[idx]
    res=findMin(l,idx+1)
    if res<minn:
        return res
    else:
        return minn
l=[12,15,1,6,11,32]
a=findMin(l,0)

print(a)
```

```python
def BS(l,i,j,key):
    if i > j:
        return -1
    mid=(i+j)//2
    if l[mid]==key:
        return mid
    elif l[mid]<key:
        i=mid+1
        res =BS(l,i,j,key)
        return res# Search in the right half
    else:
        j=mid-1
        res =BS(l,i,j,key)
        return res # Search in the left half

l=list(map(int,input().split(" ")))
key =int(input())
i = 0
j=len(l)-1
print(BS(l,i,j,key))
```
